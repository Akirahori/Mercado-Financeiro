```python
import vizro.plotly.express as px
from vizro import Vizro
import vizro.models as vm
import yfinance as yf
import datetime
```

```python
# Reinicia o ambiente do Vizro
Vizro._reset()
```

```python
# Define o ticker da Apple
ticker = "AAPL"
```

```python
# Define a data de início e fim para o último mês
data_fim = datetime.datetime.now()
data_inicio = data_fim - datetime.timedelta(days=30)
```

```python
# Obtém os dados do Yahoo Finance
dados = yf.download(ticker, start=data_inicio, end=data_fim)
```

```python
# Cria um DataFrame com os dados de cotação
df = dados.reset_index()
```

```python
df
```

```python
# Cria o gráfico de linha para as cotações de fechamento diárias
line_chart = px.line(df, x="Date", y="Close", title="Cotações de Fechamento Diárias da AAPL")
```

```python
# Cria um componente de gráfico na página do dashboard
graph_component = vm.Graph(id="line_chart", figure=line_chart)
```

```python
# Cria uma página para o dashboard
page = vm.Page(
    title="Cotações de Fechamento Diárias da AAPL",
    components=[graph_component],
)
```

```python
# Cria o dashboard com a página criada
dashboard = vm.Dashboard(pages=[page])
```

```python
# Constrói e executa o dashboard
Vizro().build(dashboard).run()
```
